# Problemas

### Problema 1

```python
valor_alvo = 100
valor_min = 0
valor_max = 50
espaco_busca = range(valor_min, valor_max + 1)
```

```python
tam_cromossomo = 5
tam_populacao = 20
num_geracoes = 100
pais_sobreviventes = 2
pais_reprodutores = 8
```

```python
selecao = "sss"
crossover = "single_point"
mutacao = "random"
taxa_mutacao = 20
```

```python
from typing_extensions import Any

def fitness(ga_instance, solution, solution_idx) -> Any:
    soma = np.sum(solution)
    erro = abs(valor_alvo - soma)
    return 1 / (erro+1)
```

```python
ag1 = pygad.GA(
    gene_space=espaco_busca,
    gene_type=int,
    num_genes=tam_cromossomo,
    sol_per_pop=pais_reprodutores,
    fitness_func=fitness,
    parent_selection_type=selecao,
    crossover_type=crossover,
    mutation_type=mutacao,
    mutation_percent_genes=taxa_mutacao,
    keep_parents=pais_sobreviventes,
    num_parents_mating=pais_reprodutores,
    num_generations=num_geracoes
)

ag1.run()
melhor_solucao, fitness_melhor_solucao, solucao_idx = ag1.best_solution()
soma_final = np.sum(melhor_solucao)
erro_final =abs(valor_alvo - soma_final)

print("Resultados Problema 1")
print(f"Melhor solução: {melhor_solucao}")
print(f"Fitness melhor solução: {fitness_melhor_solucao}")
print(f"Soma final: {soma_final}")
print(f"Erro final: {erro_final}")
```

### Problema 2

```python
pesos = np.array([
    10, 20, 30, 5, 15, 25
])
valores = np.array([
    60, 100, 120, 30, 80, 90
])
capacidade_mochila = 50

espaco_busca = [0, 1]
tam_cromossomo = 6
```

```python
from typing_extensions import Any

def fitness_2(ga_instance, solution, solution_idx) -> Any:
    peso_total = np.sum(solution * pesos)
    valor_total = np.sum(solution * valores)
    return 0 if peso_total > capacidade_mochila else valor_total
```

```python
ag2 = pygad.GA(
    gene_space=espaco_busca,
    gene_type=int,
    num_genes=tam_cromossomo,
    sol_per_pop=pais_reprodutores,
    fitness_func=fitness_2,
    parent_selection_type=selecao,
    crossover_type=crossover,
    mutation_type=mutacao,
    mutation_percent_genes=taxa_mutacao,
    keep_parents=pais_sobreviventes,
    num_parents_mating=pais_reprodutores,
    num_generations=num_geracoes
)

ag2.run()
melhor_solucao, fitness_melhor_solucao, solucao_idx = ag2.best_solution()
peso_total = np.sum(melhor_solucao * pesos)
valor_total = np.sum(melhor_solucao * valores)
itens_escolhidos = np.where(melhor_solucao == 1)[0]

print("Resultados Problema 1")
print(f"Melhor solução: {melhor_solucao}")
print(f"Fitness melhor solução: {fitness_melhor_solucao}")
print(f"Peso total: {peso_total}")
print(f"Valor total: {valor_total}")
```

### Problema 3

```python
distancias = np.array([
    [0,2,9,10,7],
    [2,0,6,4,3],
    [9,6,0,8,5],
    [10,4,8,0,6],
    [7,3,5,6,0],
])
```

```python
nomes_cidades = ["A", "B", "C", "D", "E"]
quantidade_cidades = 5
espaco_busca = list(range(quantidade_cidades))
```

```python
tam_cromossomo = quantidade_cidades
tam_populacao = 30
num_geracoes = 200
pais_sobreviventes = 2
pais_reprodutores = 10
```

```python
genes_duplicados = False
```

```python
from typing_extensions import Any

def distancia(rota) -> int:
    distancia_total = 0
    for i in range(len(rota) - 1):
        cidade_atual = rota[i]
        prox_cidade = rota[i+1]
        distancia_total += distancias[cidade_atual][prox_cidade]

    cidade_inicio = rota[0]
    cidade_fim = rota[-1]
    distancia_total += distancias[cidade_inicio][cidade_fim]

    return distancia_total

def fitness_3(ga_instance, solution, solution_idx) -> Any:
    if len(set(solution)) != quantidade_cidades:
        return 0
    return 1 / distancia(solution)
```

```python
ag3 = pygad.GA(
    gene_space=espaco_busca,
    gene_type=int,
    num_genes=tam_cromossomo,
    sol_per_pop=pais_reprodutores,
    fitness_func=fitness_3,
    parent_selection_type=selecao,
    crossover_type=crossover,
    mutation_type=mutacao,
    mutation_percent_genes=taxa_mutacao,
    keep_parents=pais_sobreviventes,
    num_parents_mating=pais_reprodutores,
    num_generations=num_geracoes,
    allow_duplicate_genes=False
)

ag3.run()
melhor_solucao, fitness_melhor_solucao, solucao_idx = ag3.best_solution()

print("Resultados Problema 1")
print(f"Distância melhor solução: {distancia(melhor_solucao)}")
print(f"Rota melhor solução: {[nomes_cidades[i] for i in melhor_solucao]}")
print(f"Fitness melhor solução: {fitness_melhor_solucao}")
```

```python
entrada_names = ("Salada Simples", "Sopa de Legumes", "Pão de alho", "Bruschetta")
pratos_names = (
    "Frango grelhado",
    "Macarrão",
    "Peixe Assado",
    "Hamburguer"
)
bebidas_names= (
    "Água",
    "Suco natural",
    "Refrigerante",
    "Vitamina"
)
```

```python
entradas = np.array([100, 150, 200, 250])
pratos = np.array([300, 450, 350, 600])
bebidas = np.array([0, 120, 150, 250])
```

```python
meta_calorica = 1000
espaco_busca = [
    range(len(entradas)),
    range(len(pratos)),
    range(len(bebidas)),
]
```

```python
tam_cromossomo = 3
tam_populacao = 30
num_geracoes = 200
pais_sobreviventes = 2
pais_reprodutores = 10
```

```python
from typing_extensions import Any

def fitness_4(ga_instance, solution, solution_idx) -> Any:
    entrada_idx = int(solution[0])
    prato_idx = int(solution[1])
    bebida_idx = int(solution[2])

    total_refeicao = (
        entradas[entrada_idx]
        + pratos[prato_idx]
        + bebidas[bebida_idx]
    )

    erro = abs(total_refeicao - meta_calorica)
    return 1 / (erro + 1)
```

```python
taxa_mutacao = 40

ag4 = pygad.GA(
    gene_space=espaco_busca,
    gene_type=int,
    num_genes=tam_cromossomo,
    sol_per_pop=pais_reprodutores,
    fitness_func=fitness_4,
    parent_selection_type=selecao,
    crossover_type=crossover,
    mutation_type=mutacao,
    mutation_percent_genes=taxa_mutacao,
    keep_parents=pais_sobreviventes,
    num_parents_mating=pais_reprodutores,
    num_generations=num_geracoes,
)

ag4.run()
melhor_solucao, fitness_melhor_solucao, solucao_idx = ag4.best_solution()

itens_escolhidos = ([
    entrada_names[melhor_solucao[0]],
    pratos_names[melhor_solucao[1]],
    bebidas_names[melhor_solucao[2]]
])

soma_final = (
    entradas[melhor_solucao[0]] +
    pratos[melhor_solucao[1]] +
    bebidas[melhor_solucao[2]]
)

print("Resultados Exercício 1")
print(f"Melhor solução: {melhor_solucao}")
print(f"""Itens escolhidos: {itens_escolhidos}""")
print(f"""Total de Calorias: {soma_final}""")
print(f"Erro: {abs(meta_calorica - soma_final)}")
print(f"Fitness melhor solução: {fitness_melhor_solucao}")
```

### Exercício 2

1. A distribuição ficou equilibrada entre as máquinas? <br>
R: Sim.

2. Qual máquina ficou com maior carga? <br>
R: A máquina 3.

3. O que acontece ao aumentar o número de gerações? <br>
R: O problema ficou perfeitamente balanceado (média das duracões).

4. O que acontece ao aumentar o tamanho da população? <br>
R: Mesma coisa que a pergunta 3, ficou perfeitamente balanceado.

5. Por que esse problema é um problema de minimização? <br>
R: Pois este problema requer que seja achado a melhor alocacão de tempo para cada máquina, de maneira que a distruicão entre elas fique balanceada. Dessa forma, uma máquina não terá muito mais tempo que outra.

6. Como o fitness transforma a minimização em maximização? <br>
R: Geralmente as os algoritimos genéticos buscam maximar algo, mas como também existem os problemas de minimizacão, o objetivo passa a ser traduzir o menor valor do problema para o maior fitness. O jeito que foi feito aqui é usando o número recíproco.


```python
duracao = np.array([
    4, 8, 6, 3, 7, 2, 5, 9, 4, 6
])

maquinas = np.array([1, 2, 3])
```

```python
espaco_busca = [range(len(maquinas)) for _ in range(len(duracao))]
```

```python
tam_cromossomo = len(duracao)
tam_populacao = 100
num_geracoes = 250
pais_sobreviventes = 2
pais_reprodutores = 10
```

```python
from typing_extensions import Any
from math import floor

avg_duracao_por_maquina = floor(np.sum(duracao) / len(maquinas))

def duracao_total_cada_maquina(solution) -> list:
    duracao_total = []
    for maquina in range(len(maquinas)):
        duracao_total.append(
            sum([
                duracao[i] if solution[i] == maquina
                else 0.0 for i in range(len(solution))
            ])
        )
    return duracao_total

def error_func_avg_duracao(duracao_total: list) -> float:
    errors_sum = [
        abs(avg_duracao - avg_duracao_por_maquina)
        for avg_duracao in duracao_total
    ]
    return 1 / (1 + sum(errors_sum))

def fitness_5(ga_instance, solution, solution_idx) -> Any:
    # cada máquina
    duracao_total = duracao_total_cada_maquina(solution)
    return error_func_avg_duracao(duracao_total)
```

```python
taxa_mutacao = 40

ag5 = pygad.GA(
    gene_space=espaco_busca,
    gene_type=int,
    num_genes=tam_cromossomo,
    sol_per_pop=pais_reprodutores,
    fitness_func=fitness_5,
    parent_selection_type=selecao,
    crossover_type=crossover,
    mutation_type=mutacao,
    mutation_percent_genes=taxa_mutacao,
    keep_parents=pais_sobreviventes,
    num_parents_mating=pais_reprodutores,
    num_generations=num_geracoes,
)

ag5.run()
melhor_solucao, fitness_melhor_solucao, solucao_idx = ag5.best_solution()

tarefas_por_maquina = {
    maquinas[maquina]: tarefa for tarefa, maquina in enumerate(melhor_solucao)
}

duracao_total_melhor_solucao = duracao_total_cada_maquina(melhor_solucao)

print("Resultados Exercício 2")
print(f"Melhor solução: {melhor_solucao}")
print(f"Tarefas por máquina: {tarefas_por_maquina}")
print(f"Tempo total de cada máquina: {duracao_total_melhor_solucao}")
print(f"Maior tempo encontrado: {max(duracao_total_melhor_solucao)}")
print(f"Fitness melhor solução: {fitness_melhor_solucao}")
```

### Exercício 3

```python
nome_aulas = [
    "Matématica",
    "Programação",
    "Fisíca",
    "Banco de Dados",
    "História",
    "Redes",
    "Inglês",
    "Estatística"
]

horarios = [
    "Segunda 08h",
    "Segunda 10h",
    "Terça 08h",
    "Terça 10h",
    "Quarta 08h"
]

# NOTE
# id | capacidade | tem projetor? | tem lab?
salas = np.array([
    [1, 30, 1, 0],
    [2, 50, 1, 0],
    [3, 25, 0, 1],
    [4, 40, 1, 1],
    [5, 50, 1, 1]
])

# NOTE
# qntd alunos | horário | precisa projetor? | precisa lab?
aulas = np.array([
    [35, 0, 1, 0],
    [25, 0, 1, 1],
    [30, 1, 1, 0],
    [40, 2, 1, 1],
    [20, 3, 0, 0],
    [35, 2, 1, 1],
    [28, 1, 1, 0],
    [45, 4, 1, 0],
])
```

```python
espaco_busca = [
    range(len(salas)) for _ in range(len(aulas))
]
```

```python
tam_cromossomo = len(aulas)
tam_populacao = 30
num_geracoes = 500
pais_sobreviventes = 2
pais_reprodutores = 10
```

```python
from typing_extensions import Any

def verificar_laboratorio(solution) -> bool:
    for aula, sala in enumerate(solution):
        precisa_lab = aulas[aula][2]
        sala_tem_lab = salas[sala][2]

        if precisa_lab and not sala_tem_lab:
            return False
    return True

def verificar_projetor(solution) -> bool:
    for aula, sala in enumerate(solution):
        precisa_proj = aulas[aula][3]
        sala_tem_proj = salas[sala][3]

        if precisa_proj and not sala_tem_proj:
            return False
    return True

def verificar_capacidade(solution) -> bool:
    for aula, sala in enumerate(solution):
        if aulas[aula][0] > salas[sala][1]:
            return False
    return True

def verificar_horarios(solution) -> bool:
    ocupacao = {}
    for aula, sala in enumerate(solution):
        horario = aulas[aula][1]

        if horario not in ocupacao:
            ocupacao[horario] = []

        if sala in ocupacao[horario]:
            return False

        ocupacao[horario].append(sala)
    return True

def calcular_desperdicio_total(solution) -> int:
    desperdicio_total = 0
    for aula, sala in enumerate(solution):
        alunos = aulas[aula][0]
        capacidade = salas[sala][1]

        desperdicio_total += capacidade - alunos
    return desperdicio_total

def fitness_6(ga_instance, solution, solution_idx) -> Any:
    penalidades = 0

    if not verificar_laboratorio(solution):
        penalidades += 1000

    if not verificar_projetor(solution):
        penalidades += 1000

    if not verificar_capacidade(solution):
        penalidades += 1000

    if not verificar_horarios(solution):
        penalidades += 1000

    penalidades += calcular_desperdicio_total(solution)

    return 1 / (1 + penalidades)
```

```python
def sala_atribuida_aula(solution) -> dict[str, list[str]]:
    res = {}
    for aula, sala in enumerate(solution):
        nome_aula =  nome_aulas[aula]
        sala = f"Sala {salas[sala][0]}"
        if not sala in res:
            res[sala] = []
        res[sala].append(nome_aula)
    return res

def calc_conflitos(solution) -> int:
    ocupacao = {}
    conflitos = 0
    for aula, sala in enumerate(solution):
        horario = aulas[aula][1]

        if horario not in ocupacao:
            ocupacao[horario] = []

        if sala in ocupacao[horario]:
            conflitos += 1

        ocupacao[horario].append(sala)
    return conflitos

def calc_restricoes_violadas(solution) -> int:
    restricoes_violadas = 0

    if not verificar_laboratorio(solution):
        restricoes_violadas += 1

    if not verificar_projetor(solution):
        restricoes_violadas += 1

    if not verificar_capacidade(solution):
        restricoes_violadas += 1

    if not verificar_horarios(solution):
        restricoes_violadas += 1

    return restricoes_violadas
```

```python
taxa_mutacao = 40

ag6 = pygad.GA(
    gene_space=espaco_busca,
    gene_type=int,
    num_genes=tam_cromossomo,
    sol_per_pop=pais_reprodutores,
    fitness_func=fitness_6,
    parent_selection_type=selecao,
    crossover_type=crossover,
    mutation_type=mutacao,
    mutation_percent_genes=taxa_mutacao,
    keep_parents=pais_sobreviventes,
    num_parents_mating=pais_reprodutores,
    num_generations=num_geracoes,
)

ag6.run()
melhor_solucao, fitness_melhor_solucao, solucao_idx = ag6.best_solution()

# NOTE
# Isso se torna um problema de minizacão
# pelo fato de estarmos buscando a menor quantidade
# de conflitos de horário.

print("Resultados Exercício 3")
print(f"Melhor solução: {melhor_solucao}")
print(f"A sala atribuída a cada aula: {sala_atribuida_aula(melhor_solucao)}")
print(f"Quantidade de conflitos: {calc_conflitos(melhor_solucao)}")
print(f"Quantidade de restricões violadas: {calc_restricoes_violadas(melhor_solucao)}")
print(f"Fitness melhor solução: {fitness_melhor_solucao}")
```
